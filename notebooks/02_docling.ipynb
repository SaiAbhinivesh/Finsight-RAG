```python
import time
from finsight.ingestion.docling_loader import PARSED_DIR, build_converter
from finsight.ingestion.loaders import RAW_DIR

converter = build_converter()
start = time.perf_counter()
result = converter.convert(str(RAW_DIR / "3M_2018_10K.pdf"))
elapsed = time.perf_counter() - start
doc = result.document
print(f"{len(doc.pages)} pages in {elapsed:.0f}s ({elapsed / len(doc.pages):.2f}s per page)")

PARSED_DIR.mkdir(parents=True, exist_ok=True)
doc.save_as_json(PARSED_DIR / "3M_2018_10K.json")
```

```python
md = doc.export_to_markdown()
i = md.find("Purchases of property, plant and equipment")
print(md[i - 1500 : i + 800])
```
